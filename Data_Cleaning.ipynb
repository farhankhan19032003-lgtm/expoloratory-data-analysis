# ***FUNCTION TRANSFORMER***

```python
dataset = pd.read_csv(r"/loan.csv")
dataset
```

```python
dataset["Loan_amount"].fillna(dataset["Loan_amount"].mean(0),inplace=True)
```

```python
dataset.isnull().sum()
```

```python
sns.displot(dataset["Loan_amount"])
plt.show()
```

```python
q1 = dataset["Loan_amount"].quantile(0.25)
q3 = dataset["Loan_amount"].quantile(0.75)
iqr = q3-q1
```

```python
min_r = q1-(1.5*iqr)
max_r = q3+(1.5*iqr)
min_r,max_r
```

```python
#dataset = dataset[dataset["Loan_amount"]<=max_r]
```

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
ft = FunctionTransformer(func=np.log1p)
```

```python
ft.fit(dataset[["Loan_amount"]])
```

```python
dataset["Loan_amount_tf"] = ft.transform(dataset[["Loan_amount"]])
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
sns.displot(dataset["Loan_amount"])
plt.title("Befor")

plt.subplot(1,2,2)
sns.displot(dataset["Loan_amount_tf"])
plt.title("After")
plt.show()
```
