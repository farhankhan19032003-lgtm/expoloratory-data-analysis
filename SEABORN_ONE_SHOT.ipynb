```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
var=sns.load_dataset("tips")
var
```

```python
sns.boxplot(y=var["size"])
plt.show()
```

```python
sns.boxplot(x=var["size"])
plt.show()
```

```python
sns.barplot(x="day",y="total_bill",data=var)
plt.show()
```

```python
sns.barplot(x="day",y="total_bill",data=var)
plt.grid()
plt.show()
```

```python
sns.set_style("darkgrid")
sns.barplot(x="day",y="total_bill",data=var,palette="Oranges")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(x="day",y="total_bill",data=var,palette="Oranges")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(x="day",y="total_bill",data=var,palette="Oranges")

sns.despine()
plt.show()
```

```python
sns.set_style("whitegrid")

plt.figure(figsize=(10,5))

sns.barplot(x="day",y="total_bill",data=var,palette="Oranges")

sns.despine()
plt.show()
```

```python
sns.set_style("white")

sns.set_context("poster",font_scale=2)

sns.barplot(x="day",y="total_bill",data=var,palette="Accent")

plt.show()
```

```python
sns.set_style("white")

sns.set_context("talk",font_scale=2)

sns.barplot(x="day",y="total_bill",data=var,palette="PuRd_r")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
var=sns.load_dataset("tips")
var
```

```python
fg = sns.FacetGrid(var)

fg.map(plt.scatter,"total_bill","tip")

plt.show()
```

```python
fg = sns.FacetGrid(var,col="sex",hue="day")

fg.map(plt.scatter,"total_bill","tip")

plt.show()
```

```python
fg = sns.FacetGrid(var,col="sex",hue="day")

fg.map(plt.scatter,"total_bill","tip").add_legend()

plt.show()
```

```python
fg = sns.FacetGrid(var,col="day",hue="sex")

fg.map(plt.scatter,"total_bill","tip").add_legend()

plt.show()
```

```python
fg = sns.FacetGrid(var,col="day",hue="sex")

fg.map(plt.bar,"total_bill","tip").add_legend()

plt.show()
```

```python
fg = sns.FacetGrid(var,col="sex",hue="day")

fg.map(plt.bar,"total_bill","tip").add_legend()

plt.show()
```

```python
fg = sns.FacetGrid(var,col="sex",hue="day",palette='summer')

fg.map(plt.bar,"total_bill","tip").add_legend()

plt.show()
```

```python
fg = sns.FacetGrid(var,col="sex",hue="day",palette='summer')

fg.map(plt.bar,"total_bill","tip",edgecolor="r").add_legend()

plt.show()
```
